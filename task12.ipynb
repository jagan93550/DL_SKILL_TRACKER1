```python
# TASK 12: Batch Normalization

import torch
import torch.nn as nn
import torch.optim as optim

class ModelWithBatchNorm(nn.Module):
    def __init__(self):
        super(ModelWithBatchNorm, self).__init__()
        self.fc1 = nn.Linear(10, 50)
        self.bn1 = nn.BatchNorm1d(50)
        self.fc2 = nn.Linear(50, 50)
        self.bn2 = nn.BatchNorm1d(50)
        self.fc3 = nn.Linear(50, 1)
    
    def forward(self, x):
        x = self.fc1(x)
        x = self.bn1(x)
        x = torch.relu(x)
        
        x = self.fc2(x)
        x = self.bn2(x)
        x = torch.relu(x)
        
        x = self.fc3(x)
        return x

# Model without batch norm for comparison
class ModelWithoutBatchNorm(nn.Module):
    def __init__(self):
        super(ModelWithoutBatchNorm, self).__init__()
        self.fc1 = nn.Linear(10, 50)
        self.fc2 = nn.Linear(50, 50)
        self.fc3 = nn.Linear(50, 1)
    
    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = self.fc3(x)
        return x

# Create models
model_bn = ModelWithBatchNorm()
model_no_bn = ModelWithoutBatchNorm()

# Test data
X = torch.randn(32, 10)
y = torch.randn(32, 1)

# Forward pass comparison
output_bn = model_bn(X)
output_no_bn = model_no_bn(X)

print(f"Model with BatchNorm output shape: {output_bn.shape}")
print(f"Model without BatchNorm output shape: {output_no_bn.shape}")
print("Batch normalization implemented successfully!")
```
