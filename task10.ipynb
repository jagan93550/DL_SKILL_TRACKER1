```python
# TASK 10: Model Fine-tuning

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.models as models

# Load pretrained model
model = models.resnet18(pretrained=True)

# Strategy 1: Freeze early layers, fine-tune later layers
for i, param in enumerate(model.parameters()):
    if i < 50:  # Freeze first 50 parameters
        param.requires_grad = False
    else:
        param.requires_grad = True

# Replace final layer
model.fc = nn.Linear(512, 10)

# Use different learning rates for different layers
param_groups = [
    {'params': [p for name, p in model.named_parameters() if 'layer3' in name or 'layer4' in name], 
     'lr': 0.001},
    {'params': [p for name, p in model.named_parameters() if 'fc' in name], 
     'lr': 0.01}
]

optimizer = optim.SGD(param_groups)

print("Fine-tuning setup complete!")
print(f"Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad)}")
print(f"Frozen parameters: {sum(p.numel() for p in model.parameters() if not p.requires_grad)}")
```
