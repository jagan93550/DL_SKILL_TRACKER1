```python
# TASK 9: Transfer Learning

import torch
import torch.nn as nn
import torchvision.models as models

# Load pretrained ResNet18
model = models.resnet18(pretrained=True)

# Freeze all layers
for param in model.parameters():
    param.requires_grad = False

# Replace the final layer for custom task
num_features = model.fc.in_features
model.fc = nn.Linear(num_features, 10)  # 10 classes for our task

# Only the new layer will be trained
for param in model.fc.parameters():
    param.requires_grad = True

print("Model transferred successfully!")
print(f"Number of trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad)}")

# Test forward pass
x = torch.randn(2, 3, 224, 224)
output = model(x)
print(f"Input shape: {x.shape}")
print(f"Output shape: {output.shape}")
print("Transfer learning setup complete!")
```
